```python
""" Correlation """
numeric_data = data.select_dtypes(include=np.number)
correlation_matrix = numeric_data.corr(method='pearson', min_periods=1)
correlation_matrix
```

```python
numeric_data = data.select_dtypes(include=np.number)
corr = numeric_data.corr()

plt.figure(figsize = [7, 3])
sns.heatmap(corr, cmap = 'Purples', annot = True)
```

```python
"""Boxplot with Hue parameter"""

plt.figure(figsize = [11, 5])
sns.boxplot(data = data, x = 'Type 1', y = 'Attack', hue = 'Legendary')
```

```python
from scipy.stats import probplot
```

```python
fig, axes = plt.subplots(2,2, figsize=[10, 10])

probplot(data['Speed'], plot=axes[0, 0])
axes[0, 0].set_title('Speed')
probplot(data['Attack'], plot=axes[0, 1])
axes[0, 1].set_title('Attack')
probplot(data['Defense'], plot=axes[1, 0])
axes[1, 0].set_title('Defense')
plt.show()
```

```python
train = pd.read_csv(basicpath + 'biketrain.csv', parse_dates=['datetime'])
```

```python
train.head()
```

```python
train['datetime-year'] = train['datetime'].dt.year
train['datetime-month'] = train['datetime'].dt.month
train['datetime-day'] = train['datetime'].dt.day
train['datetime-hour'] = train['datetime'].dt.hour
train['datetime-minute'] = train['datetime'].dt.minute
train['datetime-second'] = train['datetime'].dt.second
```

```python
train['datetime'].dt.month
```

```python
train.head()
```

```python
# 24시간 동안 자전거 count의 변화
x = train[:24]['datetime-hour']
y = train[:24]['count']

plt.plot(x, y)
```

```python

#24시간 동안 각 시각별로 전체 자전거 대여 횟수 평균의 변화
sns.pointplot(data = train, x = 'datetime-hour', y = 'count')
```

```python
df = pd.read_csv('/content/drive/MyDrive/biketrain.csv')

df['hour'] = pd.to_datetime(df['datetime']).dt.hour

weekday = df[df['workingday'] == 1].groupby('hour')['count'].mean()
weekend = df[df['workingday'] == 0].groupby('hour')['count'].mean()

plt.figure(figsize=(10,5))

plt.plot(weekday, label='Weekday')
plt.plot(weekend, label='Weekend')

plt.xlabel('Hour')
plt.ylabel('Rental Count')
plt.title('Bike Rentals by Hour')

plt.legend()
plt.show()
```
